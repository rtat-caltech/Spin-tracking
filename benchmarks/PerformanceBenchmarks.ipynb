```python
physics = {'B0' : '5e-6, 0, 0',
           'L' : '1.0, 1.0, 1.0',
           'yi' : '0.0, 0.0, 1.0',
           'm' : '1.20239e-26',
           't0' : '0.0',
           'tf' : '1.0',
           'T' : '0.25',
           'gamma' : '-2.038e8',
           'w' : '6.28318530718e3',
           'diffuse' : '0.0',
           'gas_coll' : 'true',
           'gravity' : 'false',
           'dist' : 'M'}

integration = {'numPerGPUBlock' : '256',
               'ioutInt' : '0.1',
               'integratorType' : '6',
               'seed' : 'rand',
               'h': '1e-4'}

scratch_dir = '/storage/home/hcodaman1/rtat3/scratch'
```

```python
if os.path.isdir('/usr/local/pace-apps/lmod/lmod/init/'):
    # Specifically for the PACE cluster
    sys.path.append('/usr/local/pace-apps/lmod/lmod/init/')
    from env_modules_python import module
    module('load', 'gcc', 'openmpi', 'hdf5')
```

```python
perf_folder = 'perf_benchmark_2'
```

```python
for _ in range(9):
    pb = benchmarks.PerformanceBenchmark(physics=physics, integration=integration,
                                         integrators=[0, 6, 7], n_range=np.power(2, np.arange(10, 17)), a_range=[0, 3e-5],
                                         executable='../build/cuda/main', scratch_dir=os.path.join(scratch_dir, perf_folder))
    pb.run()
```

```python
pb = benchmarks.PerformanceBenchmark()
data = pb.analyze(files=glob.glob(os.path.join(scratch_dir, perf_folder, 'out_*.hdf5')))
```

```python
mask = (data['a'] > 0)
data[mask].sort_values(['n', 'iType'])
```

```python
plt.rcParams.update({'font.size': 18})
```

```python
plt.figure(figsize=(10, 8))
ax = plt.gca()
for iType in set(data[mask]['iType']):
    ns = []
    times = []
    errs = []
    for npart in set(data[mask]['n']):
        d = (data[mask & (data['iType'] == iType) & (data['n'] == npart)])
        times.append(np.mean(d['sim time']))
        errs.append(np.std(d['sim time']))
        ns.append(npart)
        
    ax.errorbar(ns, times, yerr=errs, label=benchmarks.integrator_to_string(iType), marker='o', linestyle='')
ax.legend()
ax.grid(which='minor')
ax.grid(which='major', linewidth=3)
ax.set_ylabel('Runtime [ms]')
ax.set_xlabel('Number of Particles')
ax.set_yscale('log')
ax.set_xscale('log')
plt.savefig('performance_comparison.eps')
```
